```python
import os, subprocess

DATA_DIR = "data/scbloodnl"
os.makedirs(DATA_DIR, exist_ok=True)
BASE = "https://molgenis26.gcc.rug.nl/downloads/1m-scbloodnl/"

FILES = [
    "10x_v2_RNA_matrix.mtx.gz", "10x_v2_RNA_features.tsv.gz", "10x_v2_barcodes.tsv.gz",
    "10x_v3_RNA_matrix.mtx.gz", "10x_v3_RNA_features.tsv.gz", "10x_v3_barcodes.tsv.gz",
    "1M_cell_types.tsv",
    "1M_assignments_conditions_expid.tsv",
]

for f in FILES:
    out = os.path.join(DATA_DIR, f)
    print(f"\n=== {f}")
    # -C - resumes an interrupted download; already-complete files are skipped by curl
    subprocess.run(["curl", "-L", "-C", "-", "--retry", "5", "--progress-bar", "-o", out, BASE + f])
    print(f"{f}: {os.path.getsize(out) / 1e9:.2f} GB")
```

```python
import pandas as pd

ct = pd.read_csv(f"{DATA_DIR}/1M_cell_types.tsv", sep="\t", nrows=5)
asg = pd.read_csv(f"{DATA_DIR}/1M_assignments_conditions_expid.tsv", sep="\t")
print("cell types file:\n", ct, "\n")
print("assignments file:\n", asg.head(), "\n")
print("columns:", asg.columns.tolist())
for col in asg.columns:
    if asg[col].nunique() < 50:
        print(f"\n{col}: {asg[col].nunique()} values ->", asg[col].value_counts().head(10).to_dict())
    else:
        print(f"\n{col}: {asg[col].nunique()} unique values")
```
